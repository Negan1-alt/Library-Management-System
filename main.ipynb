```python
import sqlite3
connection = sqlite3.connect('library.db')
```

```python
from books import Book

mybook = Book(1,"Clean Code", "Robert Martin", "Programming", 5, 10)
```

```python
cursor = connection.cursor()
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS users(
    user_id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    email TEXT UNIQUE NOT NULL
)
""")
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS books(
    book_id INTEGER PRIMARY KEY,
    title TEXT NOT NULL,
    author TEXT NOT NULL,
    category TEXT NOT NULL,
    copies INTEGER NOT NULL
)
""")
```

```python
cursor.execute("""
INSERT INTO books 
VALUES (?,?,?,?,?)
""",
               (mybook.book_id, mybook.title, mybook.author, mybook.category, mybook.copies))
connection.commit()
```

```python
cursor.execute("DROP TABLE IF EXISTS books")
```

```python
cursor.execute("SELECT * FROM books")
all_books = cursor.fetchall()

for book in all_books:

    print(f"ID: {book[0]}, Title: {book[1]}, Author:{book[2]}, Category: {book[3]}, Copies: {book[4]}")
```
